```julia
# 1. Обычный алгоритм Евклида
function euclid_gcd(a, b)
    a, b = abs(a), abs(b)
    while b != 0
        a, b = b, a % b
    end
    return a
end

# 2. Бинарный алгоритм Евклида
function binary_gcd(a, b)
    a, b = abs(a), abs(b)
    g = 1
    while iseven(a) && iseven(b)
        a ÷= 2
        b ÷= 2
        g *= 2
    end
    u, v = a, b
    while u != 0
        while iseven(u)
            u ÷= 2
        end
        while iseven(v)
            v ÷= 2
        end
        if u >= v
            u -= v
        else
            v -= u
        end
    end
    return g * v
end

# 3. Расширенный алгоритм Евклида
function extended_euclid(a, b)
    a, b = abs(a), abs(b)
    r_prev, r = a, b
    x_prev, x = 1, 0
    y_prev, y = 0, 1
    while r != 0
        q = r_prev ÷ r
        r_prev, r = r, r_prev - q * r
        x_prev, x = x, x_prev - q * x
        y_prev, y = y, y_prev - q * y
    end
    return (r_prev, x_prev, y_prev)
end

# 4. Расширенный бинарный алгоритм Евклида
function extended_binary_gcd(a, b)
    a, b = abs(a), abs(b)
    g = 1
    while iseven(a) && iseven(b)
        a ÷= 2
        b ÷= 2
        g *= 2
    end
    u, v = a, b
    A, B, C, D = 1, 0, 0, 1
    while u != 0
        while iseven(u)
            u ÷= 2
            if iseven(A) && iseven(B)
                A ÷= 2
                B ÷= 2
            else
                A = (A + b) ÷ 2
                B = (B - a) ÷ 2
            end
        end
        while iseven(v)
            v ÷= 2
            if iseven(C) && iseven(D)
                C ÷= 2
                D ÷= 2
            else
                C = (C + b) ÷ 2
                D = (D - a) ÷ 2
            end
        end
        if u >= v
            u -= v
            A -= C
            B -= D
        else
            v -= u
            C -= A
            D -= B
        end
    end
    d = g * v
    x, y = C, D
    return (d, x, y)
end

# Примеры использования функций
a, b = 12345, 54321

println("Обычный алгоритм Евклида: НОД($a, $b) = ", euclid_gcd(a, b))
println("Бинарный алгоритм Евклида: НОД($a, $b) = ", binary_gcd(a, b))

d, x, y = extended_euclid(a, b)
println("Расширенный алгоритм Евклида: НОД($a, $b) = $d, коэффициенты: x = $x, y = $y")

d, x, y = extended_binary_gcd(a, b)
println("Расширенный бинарный алгоритм: НОД($a, $b) = $d, коэффициенты: x = $x, y = $y")
```
