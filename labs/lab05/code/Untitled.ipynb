```julia
using Random 

function jacobi_symbol(a::Int, n::Int)
    if n % 2 == 0 || n < 3
        throw(ArgumentError("n должно быть нечётным и >= 3"))
    end

    a = a % n
    g = 1

    while a != 0
        # Удаление множителей 2
        while a % 2 == 0
            a ÷= 2
            if n % 8 == 3 || n % 8 == 5
                g = -g
            end
        end

         a, n = n, a

         if a % 4 == 3 && n % 4 == 3
            g = -g
        end
        
        a = a % n
    end
    
    return n == 1 ? g : 0
end

function fermat_test(n::Int, k::Int=5)
    if n < 5
        throw(ArgumentError("n должно быть >= 5"))
    end
    
    if n % 2 == 0
        return false
    end
    
    for _ in 1:k
        a = rand(2:(n-2))
        if powermod(a, n-1, n) != 1
            return false
        end
    end
    return true
end

function solovay_strassen_test(n::Int, k::Int=5)
    if n < 5
        throw(ArgumentError("n должно быть >= 5"))
    end
    if n % 2 == 0
        return false
    end

    for _ in 1:k
        a = rand(2:(n-2))
        r = powermod(a, (n-1)÷2, n)
        
        if r != 1 && r != n-1
            return false
        end
        
        s = jacobi_symbol(a, n)
        if r % n != s % n
            return false
        end
    end
    return true
end

function miller_rabin_test(n::Int, k::Int=5)
    if n < 5
        throw(ArgumentError("n должно быть >= 5"))
    end
    if n % 2 == 0
        return false
    end

    s = 0
    r = n - 1
    while r % 2 == 0
        s += 1
        r ÷= 2
    end

    for _ in 1:k
        a = rand(2:(n-2))
        y = powermod(a, r, n)
        
        if y != 1 && y != n-1
            j = 1
            while j < s && y != n-1
                y = powermod(y, 2, n)
                if y == 1
                    return false
                end
                j += 1
            end
            if y != n-1
                return false
            end
        end
    end
    return true
end

numbers = [5, 7, 11, 13, 15, 17, 19, 21, 23, 29, 31, 37, 41, 43, 47, 49, 561, 1105, 1729]
println("Тест простоты для чисел: $numbers")
for n in numbers
    println("$n: Ферма=$(fermat_test(n)), Соловэй=$(solovay_strassen_test(n)), Миллер-Рабин=$(miller_rabin_test(n))")
end
```
